```python
p = india["EV_Intender"].mean() # Probability
p
```

```python
n = len(india)
n
```

```python
z = 1.96 # 95% normal confidence interval
```

```python
# calculate standard error / margin of error
se = math.sqrt(p * (1 - p) / n)
se
```

```python
# Confidence Interval
lower = p - z * se
upper = p + z * se

print(lower, upper)
```

```python
print(f"Sample Data EV Buying Intention in India: {p*100:.2f}%")
print(f"The Confidence Interval is from {lower*100:.2f}% to {upper*100:.2f}%")
```

```python
# Customer Segmentation using Rule based Scoring
```

```python
data.columns
```

```python
data[['EV_Intender', 'AI_Customization_Likelihood',
      'Home_Charger_Access', 'Preferred_Next_Engine']].head()
```

```python
def assign_segment(row):
  # EV intenders with strong AI interest are considered digitally engaged.
  if row['EV_Intender'] == 1 and row['AI_Customization_Likelihood'] >= 4:
    if row['Home_Charger_Access'] == "Yes":
      return "EV Ready"
    else:
      return "EV Interested. But Charging Constraints"

  if row['Preferred_Next_Engine'] == "HEV":
    return "Hybrid Practicality Seeker"

  return "Traditional Vehicle Buyer"
```

```python
data.apply(assign_segment, axis=1)
```

```python
data["Customer_Segment"] = data.apply(assign_segment, axis=1)
data.head()
```
